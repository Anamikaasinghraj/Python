### Inheritance:- Inheritance allows one class (child/derived class) to acquire properties and methods of another class (parent/base class)
#### types:- 1.Single Inheritance 2.Multiple Inheritance 3.Multilevel Inheritance 4.Hierarchical Inheritance 5.Hybrid Inheritance

```python
class Animal:
    def sound(self):
        print("Some generic sound")

class Dog(Animal):  # Dog inherits from Animal
    def sound(self):
        print("Bark")

d = Dog()
d.sound()
```
