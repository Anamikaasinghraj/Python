### Constructor:- A constructor is a method that runs automatically when we call a class and this constructor function will target the objects location.

### __init__method:-
#### The __init__() method is a special method (constructor) in Python that is called automatically when an object is created, It is used to initialize attributes of the object.

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print(f"Name: {self.name}, Marks: {self.marks}")

s1 = Student("Anamika", 90)
s1.display()
```
