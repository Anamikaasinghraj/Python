## oop in python

### Introduction to OOP
Object-Oriented Programming (OOP) is a programming paradigm based on the concept of objects.
An object represents real-world entities (like a car, student, employee, etc.).
Each object has attributes (data) and methods (functions).
OOP makes programs more modular, reusable, and easier to maintain.
### Key Features of OOP
1.Class and Object
2.Encapsulation
3.Inheritance
4.Polymorphism
5.Abstraction


### Class and Object
#### Class
A class is a blueprint or template to create objects.
It defines attributes (variables) and methods (functions).

#### Object
An object is an instance of a class.
It represents real-world entities.

```python
class Car:
    def __init__(self, brand, model):
        self.brand = brand  # attribute
        self.model = model  # attribute
    
    def display_info(self):  # method
        print(f"Car: {self.brand} {self.model}")

# Create objects
car1 = Car("Tesla", "Model S")
car2 = Car("BMW", "X5")

# Access methods
car1.display_info()
car2.display_info()
```
